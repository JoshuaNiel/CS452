```python
# 1 Find the movies titled "Gladiator".
res1 = movies.find({"title": "Gladiator"}, {"_id":0, "director":0, "actors":0, "country":0, "summary":0})
list(res1)

```

```python
# 2 List the distinct genre values of movies.
res2 = movies.distinct("genre")
list(res2)

```

```python
# 3 List just the titles of movies of "crime" or "drama" genre.
res3 = movies.find({"genre": { "$in": ["crime", "drama"]}}, {"_id":0, "title": 1, "year": 1, "genre": 1})
list(res3)
```

```python
# 4 Find the list of movies directed by "Hitchcock", display only title and year and sort them by year (asc).
res4 = movies.find({"director.last_name": "Hitchcock"}, {"_id": 0, "title": 1, "year": 1}).sort({"year": 1})
list(res4)
```

```python
# 5 List the movie titles where "Cotillard" played.
res5 = movies.find({"actors.last_name": "Cotillard"}, {"_id": 0, "title": 1, "year": 1})
list(res5)
```

```python
# 6 List the movie titles released between 1967 and 1995 (inclusive).
res6 = movies.find({"$and": [ {"year": {"$gte": 1967}}, {"year": {"$lte": 1995}}]}, {"_id": 0, "title": 1, "year": 1}).sort({"year": 1})
list(res6)
```

```python
# 7 List the the movies released between 1967 and 1995 (inclusive), by displaying only title, year, director’s last name sorted by year
res7 = movies.find({"$and": [ {"year": {"$gte": 1967}}, {"year": {"$lte": 1995}}]}, {"_id": 0, "title": 1, "year": 1, "director.last_name":1}).sort({"year": 1})
list(res7)
```
